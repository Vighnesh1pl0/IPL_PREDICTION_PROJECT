```python
import joblib
import pandas as pd

model_path = "../models/best_winprob_model.pkl"
model = joblib.load(model_path)

def predict_win_probability(input_dict, X_columns):
    """
    Predicts probability second innings team will win.
    input_dict → dictionary containing all required features
    X_columns → columns of X used during training
    """
    row = dict.fromkeys(X_columns, 0)

    # fill numerical inputs
    for k, v in input_dict.items():
        if k in row:
            row[k] = v

    # create DataFrame
    row_df = pd.DataFrame([row], columns=X_columns)

    # predict prob
    prob = model.predict_proba(row_df)[0][1]
    return prob
```
