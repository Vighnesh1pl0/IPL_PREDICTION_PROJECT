```python
df['dismissal_kind'] = df['dismissal_kind'].fillna('Not Out')
df['player_dismissed'] = df['player_dismissed'].fillna('No Player')
df['fielder'] = df['fielder'].fillna('No Fielder')
df['extras_type'] = df['extras_type'].fillna('No Extra')
df['bowling_team'] = df['bowling_team'].fillna('Unknown')
```

```python
cat_cols = ['batsman','non_striker','bowler','batting_team','bowling_team','extras_type','dismissal_kind']

for col in cat_cols:
    df[col] = df[col].astype('category')
```

```python
df.to_csv("clean_ball_by_ball.csv", index=False)
```
